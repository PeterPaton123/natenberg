## Stock Price Modelling

If $X_t$ is a stochastic process satisfying the SDE

$$
dX_t = \mu(t,X_t)\,dt + \sigma(t,X_t)\,dW_t,
$$

where $W_t$ is standard Brownian motion, and $f(t,X_t)\in C^{1,2}$, then Itô's lemma gives

$$
df(t,X_t) = \left(f_t + \mu f_x + \tfrac12\sigma^2 f_{xx}\right)dt + \sigma f_x\,dW_t.
$$

For risk-neutral pricing, we assume a stock price $S_t$, without dividends, follows **geometric Brownian motion**:

$$
dS_t = rS_t\,dt + \sigma S_t\,dW_t.
$$

The drift $rS_t$ represents growth at the risk-free rate $r$, proportional to the current price; movements we cannot model are represented by random fluctuations with magnitude $\sigma S_t$, also proportional to the asset price, where $\sigma$ is volatility.

## Derivation of asset price

Let $f(t,S_t)=\ln S_t$. Its partial derivatives are

$$
\frac{\partial f}{\partial t}=0,\qquad
\frac{\partial f}{\partial S}=\frac{1}{S_t},\qquad
\frac{\partial^2 f}{\partial S^2}=-\frac{1}{S_t^2}.
$$

Applying Itô's lemma,

$$
\begin{aligned}
d\ln S_t
&=\left(0+rS_t\frac{1}{S_t}
+\frac12\sigma^2S_t^2\left(-\frac{1}{S_t^2}\right)\right)dt
+\sigma S_t\frac{1}{S_t}\,dW_t\\
&=\left(r-\frac12\sigma^2\right)dt+\sigma\,dW_t.
\end{aligned}
$$

For constant $r$ and $\sigma$, integrating from $0$ to $t$ with $W_0=0$ gives

$$
\ln\frac{S_t}{S_0}=\left(r-\frac12\sigma^2\right)t+\sigma W_t,
$$

so

$$
\boxed{S_t=S_0\exp\left[\left(r-\frac12\sigma^2\right)t+\sigma W_t\right]}.
$$

Since $W_t\sim\mathcal N(0,t)$, log returns are normal with mean $(r-\tfrac12\sigma^2)t$ and variance $\sigma^2t$; hence $S_t$ is lognormal and remains positive for $S_0>0$.
