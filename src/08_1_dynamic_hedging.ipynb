# 08.1 Dynamic Hedging

Taking interest rates and underlying volatility to be constant, an option position can be replicated by taking out positions in the underlying such that the delta of the underlying position replicates the option's delta at that underlying price at that point in time. This also assumes continuous underlying price movements, continuous trading, no transaction costs, and borrowing and lending at the same interest rate, and a European option on an underlying that pays no dividends.

Instantaneously updating the underlying position as the underlying moves and time passes means the delta of the underlying position is always aligned with the corresponding option position. Assuming cash is invested at the interest rate and any borrowing is charged at that same rate, this strategy reproduces the option's actual payoff at expiry.

By no-arbitrage, with an equal pay-off at expiry, the initial net cost of entry into the underlying-and-cash position equals the option's theoretical value. Interest and opportunity costs are accounted for through the cash account and discounting, rather than added separately. After entry, purchases and sales are funded within the portfolio, without adding outside money.

With discrete updates, replication is approximate. Under these assumptions, as the updates become continuous, the replication error tends to zero.

## Capturing Theoretical Value of an Option

If we think a call is underpriced, we purchase it with the ambition of capturing the difference between its price and theoretical value. Ideally we reduce the variance of total payout. We do this by hedging deltas in the underlying. A natural tradeoff options traders face is how much you want to spen hedging deltas since more frequent hedging reduces variance by costs EV.

Intuitively, an option has extrinsic value as price fluctuations do not symmetrically affect its payoff: positive price movements can pay an unlimited amounts, negative movements can only take the call's payoff to zero. If we hedge against deltas, doesn't the extrinsic value go away? How can we cash in on the asymmetry while simultaneously hedging movements in the underlying.

This is possible due to the curvature of an option. For example, if we purchase an 100 lots of option with a delta of 60, we can short 60 lots of the underlying to be delta neutral. If the underlying moves up such that the new delta of the option is 70, we have a net +1,000 delta position and must short an additional 10 lots of the underlying. If the underlying moves back down such that the delta is 60 again, we now have a negative delta and purchase back those 10 lots. In doing this the option has lost some extinsic value (e.g. due to theta) however we have sold high and bought low.

More generally,

- If the stock rises, the call's delta increases, so we short more shares at the higher price.
- If the stock falls back, delta decreases, so we buy those extra shares back at the lower price.

That round trip generates a trading gain. This is how the option's curvature, called **gamma**, can translate stock movements into hedging cash flows. Not every path produces a profitable round trip.

There is a cost to holding this opportunity: **the option loses time value as time passes**. In the ideal model, the gains associated with curvature balance time decay and financing when we pay the theoretical price.

If we buy below the correct theoretical value and continuously hedge under the model's assumptions, we capture that pricing difference in present-value terms. 

Therefore, we reduce the variance of the combined option-and-hedge result, not the option's own payoff. If our volatility assumption is wrong, that profit is not guaranteed.

## Hedging deltas on with large Gamma positions

A funny non-obvious property of hedging deltas is that the market impact of your delta hedges hurts your vol position. Assuming you push the underlying in the direction of your delta hedges then hedging deltas while long vol decreases the realised vol and hedging deltas when short vol increases realised vol. Basically if you're long vol and it ticks up then you sell deltas which pushes it back down; short vol and it ticks up then you buy deltas which makes it tick up more. It's one of the things that limit capacity in running options positions; larger positions demand larger delta hedges which reduces the vol edge on your entire position. If a firm has a huge gamma position and they would account for this cost in deciding how often to hedge.
