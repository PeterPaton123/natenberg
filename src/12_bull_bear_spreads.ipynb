Purchasing calls and selling puts is positive delta position. Selling calls and purchasing puts is a negative delta position. Traders can use these unhedged to take positions in the underlying, as well as volatility positions. 

For example if the trader thinks an underlying is overpriced and the IV is too high they can sell a call option, if they're right the call was overpriced as priced with the IV being great as well as taking a negative delta position. Selling a call leaves the trader open to potentially unlimited losses.
