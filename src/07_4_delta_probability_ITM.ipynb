# 07.4: Delta as Probability of Finishing In the Money

**Thesis:** The magnitude of delta is a rough proxy for the probability that an option expires in the money. Delta measures how much the option's value changes for a small move in the underlying; the chance of finishing in the money helps determine how often that move will affect its payoff.

For a call, consider three cases:

- **Deep out of the money:** A small rise is unlikely to make the option valuable. Delta is near **0**, as is the probability of expiring in the money.
- **Deep in the money:** A small rise adds almost dollar for dollar to the option's value. Delta is near **1**, and the probability of expiring in the money is high.
- **Near the strike:** A small rise may determine whether the option has a payoff. Delta lies between **0 and 1**, reflecting the uncertainty.

**The 50/50 intuition:** Options at the money (or technically, with a strike at the futures price) have a delta magnitude of about **0.5**: assuming price fluctuations are symmetric and random, the probability of the underlying finishing on either side of the strike is about **50%**.

This is an intuition, not an exact identity: standard lognormal pricing does not give exactly 0.5 delta at the futures price. Delta is a sensitivity, and its probability interpretation is a pricing approximation rather than a real-world forecast. For puts, use the **absolute value** of delta.
